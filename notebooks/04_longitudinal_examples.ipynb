# Longitudinal examples
Run `saguarodata all` from the project root first. No network is used by this notebook.

```python
from pathlib import Path
import os
if not Path('configs').exists():
    os.chdir('..')
import saguarodata as sd
ds = sd.load('data/processed')
```

```python
linked = ds.observations(program='historic_plots').dropna(subset=['organism_id']).iloc[0]
print(ds.history(linked.organism_id)[['survey_year','height_raw','height_m','arm_count','status']].to_string(index=False))
print(ds.provenance(linked.observation_id).to_string(index=False))
assert ds.observations(program='census').organism_id.isna().all()
```
